```python
import pandas as pd
import numpy as np

def adult_eda_report(df):
    # Make a copy so original dataframe is not modified
    df = df.copy()

    # Replace '?' with NaN
    df.replace('?', np.nan, inplace=True)

    # Remove spaces from column names
    df.columns = df.columns.str.strip()

    # Remove spaces from string values
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()

    print("===== MISSING VALUE SUMMARY =====")
    print(df.isnull().sum())

    print("\n===== INCOME CLASS DISTRIBUTION =====")
    print(df['income'].value_counts())

    print("\n===== INCOME BY EDUCATION =====")
    print(pd.crosstab(df['education'], df['income']))

    print("\n===== INCOME BY OCCUPATION =====")
    print(pd.crosstab(df['occupation'], df['income']))

    print("\n===== MEAN HOURS-PER-WEEK BY INCOME =====")
    print(df.groupby('income')['hours-per-week'].mean())

    # Convert target income into numeric values
    income_numeric = (
        df['income']
        .astype(str)
        .str.replace('.', '', regex=False)
        .map({'<=50K': 0, '>50K': 1})
    )

    # Select numeric columns
    numeric_df = df.select_dtypes(include=np.number).copy()

    # Add encoded income
    numeric_df['income'] = income_numeric

    # Pearson correlation with income
    correlation = numeric_df.corr(method='pearson')['income']
    correlation = correlation.drop('income').abs().sort_values(ascending=False)

    print("\n===== PEARSON CORRELATION WITH INCOME =====")
    print(correlation)

    # Top 3 features
    top_3_features = correlation.head(3)

    print("\n===== TOP 3 FEATURES =====")
    print(top_3_features)

    return top_3_features
df = pd.read_csv("adult.csv")

top_features = adult_eda_report(df)

print("\nTop 3 features:")
print(top_features)
```
